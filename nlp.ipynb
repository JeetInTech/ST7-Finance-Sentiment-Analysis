```python
import re
import pandas as pd
import numpy as np
import nltk
import pickle
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Download necessary NLTK resources
nltk.download('punkt')
nltk.download('stopwords')

# Step 1: Load Dataset
df = pd.read_csv("data.csv")

# Step 2: Clean Data using Regex
def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # Remove special characters and numbers
    return text

df["cleaned_text"] = df["Sentence"].apply(clean_text)

# Step 3: Word Tokenization
df["tokenized"] = df["cleaned_text"].apply(word_tokenize)

# Step 4: Stopword Removal
stop_words = set(stopwords.words("english"))
df["filtered"] = df["tokenized"].apply(lambda words: [w for w in words if w not in stop_words])

# Step 5: Stemming
stemmer = PorterStemmer()
df["stemmed"] = df["filtered"].apply(lambda words: [stemmer.stem(w) for w in words])

# Convert back to text
df["final_text"] = df["stemmed"].apply(lambda words: " ".join(words))

# Step 6: TF-IDF Vectorization
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["final_text"])

# Save the vectorizer
with open("tfidf_vectorizer.pkl", "wb") as f:
    pickle.dump(vectorizer, f)

# Encode Sentiment Labels
df["label"] = df["Sentiment"].map({"positive": 2, "neutral": 1, "negative": 0})

# Step 7: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, df["label"], test_size=0.2, random_state=42)

# Step 8: Train ML Model (Naïve Bayes)
model = MultinomialNB()
model.fit(X_train, y_train)

# Save the model
with open("sentiment_model.pkl", "wb") as f:
    pickle.dump(model, f)

# Model Evaluation
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
